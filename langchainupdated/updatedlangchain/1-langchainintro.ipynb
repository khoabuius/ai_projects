```python
# Option C: LangChain create_agent pointed at OpenRouter
# The bug: create_agent(model="gpt-5") builds a plain ChatOpenAI aimed at real
# OpenAI with no key -> "Missing credentials". Pass a CONFIGURED model object
# instead of a string so the OpenRouter key + base_url are injected.
import os
from dotenv import dotenv_values
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent

def get_weather(city:str) -> str:
    "This function return weather for the city"
    return f"Weather in {city} is sunny."
# Ensure creds are loaded even if run out of order.
for _k, _v in dotenv_values("/Users/buik1/langchainupdated/.env").items():
    if _v is not None:
        os.environ[_k] = _v
openrouter_api_key = os.environ.get("OPENAI_API_KEY", "")
assert openrouter_api_key, "openrouter_api_key is empty - run cell 1 / check .env"

model = ChatOpenAI(
    model="openai/gpt-4o-mini",          # OpenRouter model id (e.g. openai/gpt-4o, openai/gpt-5)
    api_key=openrouter_api_key,
    base_url="https://openrouter.ai/api/v1",
    temperature=0,
)

agent = create_agent(
    model=model,                          # pass the object, NOT the string "gpt-5"
    tools=[get_weather],
    system_prompt="you are a helpful assistant",
)

result = agent.invoke({"messages": [{"role": "user", "content": "In one sentence, what is LangChain?"}]})
print(result["messages"][-1].content)
result = agent.invoke({"messages": [{"role": "user", "content": "What is the weather in East Peoria"}]})
print(result["messages"][-1].content)
```
