```python

import os
from dotenv import dotenv_values
# Ensure creds are loaded even if run out of order.
for _k, _v in dotenv_values("/Users/buik1/langchainupdated/.env").items():
    if _v is not None:
        os.environ[_k] = _v
open_api_key = os.environ.get("OPENAI_API_KEY", "")
google_api_key = os.environ.get("GOOGLE_API_KEY", "")
groq_api_key = os.environ.get("GROQ_API_KEY", "")

from langchain.chat_models import init_chat_model
# Update model to gemini-3.6-flash
model = init_chat_model(
    "openai/gpt-oss-120b",
    model_provider="groq",
    api_key=groq_api_key
)
for chunk in model.stream("Why does parrots have colorful feather"):
    print(chunk.text, end="", flush=True)

responses = model.batch(
  [
    "What is the first US President",
    "Why the airplain can fly",
    "Why the chicken cannot fly as bird"
  ]
)    
for res in responses:
  print(res)
```
